Answer the following questions regarding the `df_gdp` you created in Exercise 2.1:

  4. Sort the dataframe by Country and assign it to a new DataFrame

  6. What are the top 5 countries with smallest population?

```python
from google.colab import drive
drive.mount('/content/drive')

import numpy as np
import pandas as pd

# Define the dataset file path in Google Drive
dataset_path = '/content/drive/MyDrive/ICT110_Activity_Near/datasets/GDP(PPP)_LifeExpectancy.xlsx'

# Load the Excel file, exclude the 'Code' column, assign headers, and handle missing value markers
gdp_df = pd.read_excel(
    dataset_path,
    header=0,
    usecols=lambda column_name: column_name != 'Code',
    names=['Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'],
    na_values=['..', 'NA', 'N/A', 'null', 'None', '', ' ']
)

# Convert key metric columns to numeric data types, coercing errors to NaN
for col in ['Life_Exp', 'GDP', 'Population']:
    gdp_df[col] = pd.to_numeric(gdp_df[col], errors='coerce')

# Calculate and print dataset summary metrics
total_rows, total_cols = gdp_df.shape
print(f"1. Dimensions of the DataFrame: {total_rows} rows, {total_cols} columns")
print(f"2. Number of valid GDP values: {gdp_df['GDP'].count()}")
print(f"3. Average life expectancy: {gdp_df['Life_Exp'].mean():.2f} years")

# Sort the DataFrame alphabetically by country name
df_gdp_sorted = gdp_df.sort_values(by='Country')
print("4. DataFrame sorted by Country and assigned to 'df_gdp_sorted'.")

# Display specific filtered insights
print("\n5. Top 5 countries with the highest GDP:")
display(gdp_df.nlargest(5, 'GDP')[['Country', 'GDP']])

print("\n6. Top 5 countries with the smallest population:")
display(gdp_df.nsmallest(5, 'Population')[['Country', 'Population']])
```
